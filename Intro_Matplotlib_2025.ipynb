# Matplotlib for Visualization
## Import Packages

```python
# Conventional ways of importing the packages
import matplotlib as mpl
import matplotlib.pyplot as plt

import numpy as np
```

```python
x=np.linspace(-5,2,100)
y1 = x**3 + 5*x**2 + 10
y2 = 3*x**2 + 10*x

# Use the plt function to make plots directly:

plt.plot(x, y1, color='blue', label='y(x)')
plt.plot(x, y2, c='r', label="y'(x)")
#note about the double quotation marks above.

plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.xlim(-5,2)
plt.ylim(-10,40)

plt.legend()

#Set aspect ratio:
#use the get current axes function to retrun an Axes object.
ax = plt.gca()

ax.set_aspect(7/50*0.75)
# xlim_range/ylim_range*desired ratio

#plt.show()
#Or save the plot to a file:
#plt.savefig('myfirstplot.png',dpi=300)
```

```python
# Alternatively, use the plt.subplots function,
# to generate Figure and Axes objects.
fig, ax = plt.subplots()

ax.plot(x, y1, color='blue', label='y(x)')
ax.plot(x, y2, c='r', label="y'(x)")

ax.set_xlabel("x")
ax.set_ylabel("y")

ax.set_xlim(-5,2)
ax.set_ylim(-10,40)

#ax.set_aspect(7/50*1)

ax.legend()

#Save the plot to a file:
#fig.savefig('mysecondplot.png', dpi=300)
```

## Figure and Axes
**Note:**
There are various ways to define colors in Python. One common approach is to use the RGB or RGBA (red, green, blue, alpha) tuple of float values in a closed interval [0, 1]. A tuple is a collection of objects enclosed by (); the objects are ordered and immutable (i.e. the tuple content cannot be modified). Alpha is the transparency.

```python
fig = plt.figure(figsize=(10,4), facecolor=(0.9,0.3,0.6,0.5))

# axes coordinates as fractions of the canvas (figure) width and height
left, bottom, width, height = 0.1, 0.1, 0.6, 0.6
ax = fig.add_axes( (left,bottom,width,height), facecolor = 'white')

x= np.linspace(-2,2,1000)
y1 = np.cos(40*x)
y2 = np.exp(-x**2)

ax.plot(x, y1*y2, 'b')
ax.plot(x, y2, 'g')
ax.plot(x, -y2, 'g')

ax.set_xlabel("x")
ax.set_ylabel("y")
```

## Line Properties
We can control the following basic line properties: **color, alpha, linewidth (lw), linestyle (ls), marker, markersize, markerfacecolor, markeredgewidth, markeredgecolor.**


```python
# Create subplots
fig, axes = plt.subplots(1, 4, figsize=(16,3))

x = np.linspace(-5,5,5)

# np.ones_like() is a numpy built-in function. 
# It returns an array of ones with the same shape and type as a given array.
y = np.ones_like(x)

# linewidth
axes[0].plot(x,y, lw=1)
axes[0].plot(x,y+1, lw=2)
axes[0].plot(x,y+2, lw=3)
axes[0].plot(x,y+3, lw=4)

# linestyle
axes[1].plot(x,y, ls='-')
axes[1].plot(x,y+1, ls='--')
axes[1].plot(x,y+2, ls=':')
axes[1].plot(x,y+3, ls='-.')

# marker types
axes[2].plot(x,y, marker='+')
axes[2].plot(x,y+1, marker='o')
axes[2].plot(x,y+2, marker='*')
axes[2].plot(x,y+3, marker='s')

# marker size and color
axes[3].plot(x,y, marker='o', markeredgecolor='k', markersize=4, markerfacecolor='w')
axes[3].plot(x,y+1, marker='o', markeredgecolor='k', markersize=8, markerfacecolor='red',alpha=0.5)
axes[3].plot(x,y+2, marker='o', markeredgecolor='k', markersize=12, markerfacecolor='lightgreen')
axes[3].plot(x,y+3, marker='o', markeredgecolor='k',markeredgewidth=3, markersize=16, markerfacecolor='darkorange')
```

## Axes Properties

### Axes Labels and Titles


```python
fig, ax = plt.subplots(figsize=(8,2))

x = np.linspace(0,50,500)
y = np.sin(x)*np.exp(-x/10)

ax.plot(x, y, lw =2, c="b")

ax.set_xlabel("x", fontsize=18, family="serif", color='b')
ax.set_ylabel("f(x)", fontsize=18, family="serif", color="blue")

ax.set_title("Axis Labels and Title Example", fontsize=14, family='serif')
```

### Axes Ticks and Tick Labels

```python
fig, axes = plt.subplots(1,4, figsize=(12,3))

x = np.linspace(-2*np.pi, 2*np.pi, 101)
y = np.sin(x)

axes[0].plot(x,y)
axes[0].set_title("default ticks")

axes[1].plot(x,y)
axes[1].set_title("set_xticks")
axes[1].set_xticks([-5,0,5])
axes[1].set_yticks([-1,0,1])

axes[2].plot(x,y)
axes[2].set_title("set_major_locator")
axes[2].xaxis.set_major_locator(mpl.ticker.MaxNLocator(4))
axes[2].xaxis.set_minor_locator(mpl.ticker.MaxNLocator(8))
axes[2].set_ylim(-1,1)
axes[2].set_yticks([-1,0,1])
axes[2].yaxis.set_minor_locator(mpl.ticker.MaxNLocator(8))

axes[3].plot(x,y)
axes[3].set_title("set_xticklabels")
axes[3].set_xticks([-2*np.pi,0,2*np.pi])
axes[3].set_yticks([-1,0,1])
axes[3].set_xticklabels(['$-2\\pi$',0,r'$2\pi$'])
```

### Axes Grids

```python
fig, axes = plt.subplots(1,3,figsize=(12,4))

x = np.linspace(-2*np.pi, 2*np.pi, 101)
y = np.sin(x)

axes[0].plot(x,y)
axes[0].set_title("default grid")
axes[0].grid()

axes[1].set_title("major/minor grid")
axes[1].grid(color="blue", which="both", ls = ":", lw=0.5)

axes[2].set_title("individual x/y major/minor grid")
axes[2].grid(color="grey", which="major", axis='x', ls = "-.", lw=4)
axes[2].grid(color="grey", which="major", axis='y', ls = "-", lw=0.25)

# Set grid lines on minor ticks, if present.
# axes[2].grid(color="grey", which="minor", axis='both', ls = ":", lw=0.25)
```

### Scientific Notation for Tick Labels

```python
fig, axes = plt.subplots(1,2, figsize=(8,2))

x=np.linspace(0,1e5,100)
y = x**2

axes[0].plot(x,y)
axes[0].set_title("default labels", loc="right")

axes[1].plot(x,y)
axes[1].set_title("scientific notation labels", loc="right")

formatter = mpl.ticker.ScalarFormatter(useMathText=True)
formatter.set_scientific(True)
formatter.set_powerlimits((-1,1))
# Note: In set_powerlimits(lims), lims is a tuple (min_exp, max_exp),
# containing the powers of 10 that determine the switchover threshold. 

axes[1].xaxis.set_major_formatter(formatter)
axes[1].yaxis.set_major_formatter(formatter)
```

## Log Plots

```python
fig, axes = plt.subplots(1,3,figsize=(12,3))

x = np.linspace(0,1e3,100)
y1, y2 = x**3, x**4

axes[0].set_title('loglog')
axes[0].loglog(x,y1,'b', x, y2, 'r')

axes[1].set_title('semilogy')
axes[1].semilogy(x,y1,'b', x, y2, 'r')

# Same effect as the loglog plot:
axes[2].set_title('plot/set_xcale/set_yscale')
axes[2].plot(x,y1,'b',x,y2,'r')
axes[2].set_xscale('log')
axes[2].set_yscale('log')
```

## Twin Axes

```python
fig, ax1 = plt.subplots(figsize=(8,4))

r = np.linspace(0,5,100)
a = 4*np.pi*r**2 # area
v = (4*np.pi/3)*r**3

ax1.plot(r,a,lw=2, c='b')

ax1.set_xlabel(r'radius ($m$)',fontsize=16)
ax1.set_ylabel(r'surface area ($m^2$)', fontsize=16, c='b')

for label in ax1.get_yticklabels():
    label.set_color("blue")
    
ax2 = ax1.twinx()
ax2.plot(r,v,lw=2,c='r')
ax2.set_ylabel(r'volume ($m^3$)', fontsize=16, c='r')

ax1.set_ylim(ax2.get_ylim()[0],ax2.get_ylim()[1])

for label in ax2.get_yticklabels():
    label.set_color("red")
```

## Colormap Plots

```python
# Importing the colormap module:
import matplotlib.cm as cm
```

### imshow
Display data as an image, i.e., on a 2D regular raster (a matrix of cells or pixels).

```python
# Create a meshgrid:
N = 64
x=np.linspace(-np.pi,np.pi,N)
y=np.linspace(-np.pi,np.pi,N)

X,Y = np.meshgrid(x,y)
Z=np.sinc(X**2+Y**2)
# Note: sinc(theta) = sin(theta)/theta

fig, axes = plt.subplots(1,4,figsize=(12,3))

axes[0].imshow(Z, cmap=cm.hot)
axes[1].imshow(Z, cmap=cm.ocean,interpolation = 'nearest')
axes[2].imshow(Z, cmap=cm.bone,interpolation = 'spline16')
axes[3].imshow(Z, cmap=cm.cool,interpolation = 'quadric')

#Check online about possible interpolation options (e.g. nearest, bicubic, quadric, etc.)
```

### contour

```python
N=64 
x=np.linspace(-np.pi,np.pi,N) 
y=np.linspace(-np.pi,np.pi,N)

X,Y = np.meshgrid(x,y)
Z=(X**2+Y**2)

fig, ax = plt.subplots()

im = ax.imshow(Z, cmap=cm.hot, extent=(-np.pi,np.pi,-np.pi,np.pi)) 
fig.colorbar(im, ax = ax)

# contour lines with numerical labels:
levels=[0,3,6,9,12]
ct = ax.contour(X,Y,Z, levels, cmap=cm.bwr)
ax.clabel(ct, fmt='%d')
#fig.colorbar(ct, ax = ax)
```

### 3D Scatter Plots

```python
#Note about the usage of a dictionary to specify the keyword:
fig, ax = plt.subplots(figsize=(4,4),subplot_kw={'projection':'3d'})

x, y, z = np.random.randn(3,100)
r = np.sqrt(x**2+y**2+z**2)

ax.scatter(x, y, z, s=30, alpha=0.7, c=r, cmap=cm.hot)
# Note: ax.scatter can take a scalar list/array for the color parameter c.

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('3D scatter plot')
```

### Plot a Scalar field on a 3D surface

```python
#Note about the usage of a dictionary to specify the keyword:
fig, axes = plt.subplots(1,3, figsize=(14,4), subplot_kw={'projection':'3d'})

# Create mesh grid:
x=np.linspace(-3,3,256)
y=np.linspace(-3,3,256)

X, Y = np.meshgrid(x,y)
Z=np.sinc(np.sqrt(X**2+Y**2))

p = axes[0].plot_surface(X,Y,Z, cmap=cm.jet,linewidth=10, antialiased=True)
fig.colorbar(p, ax=axes[0], shrink=0.5)

axes[1].plot_wireframe(X,Y,Z, rstride=10, cstride=10, color='black')

axes[2].contour(X,Y,Z, zdir='z', cmap=cm.hot, offset=0)
axes[2].contour(X,Y,Z, zdir='y', cmap=cm.binary, offset=np.pi)
# zdir: determines on which plane the 2D plot will be drawn.

# Define all axes labels and title using a single function:
def set_title_labels(ax, title):
    ax.set_title(title)
    ax.set_xlabel('x', fontsize=16)
    ax.set_ylabel('y', fontsize=16)
    ax.set_zlabel('z', fontsize=16)

set_title_labels(axes[0],'plot_surface')
set_title_labels(axes[1],'plot_wireframe')
set_title_labels(axes[2],'contour')
```
