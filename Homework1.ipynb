# 1.

```python
def fibonacci(n):
    x = 0
    y = 1

    numbers = []

    for i in range(n):
        numbers.append(x)
        z = y
        y = x + z
        x = z

    return numbers

fibonacci(20)
```

# 2.

```python
import math

def IsPrime(n):
    b = True
    for i in range(2,int(math.sqrt(n))+1):
        if n % i == 0:
            b = False

    return b

print(IsPrime(4))
print(IsPrime(23))
```

```python
def primes(n):
    primes = []
    for i in range(2,n+1):
        if IsPrime(i):
            primes.append(i)

    return primes

primes(100)
```
