## Reshaping and Resizing
These operations are very useful when we deal with numpy arrays. We often will reshape the numpy array, so that the array has a proper shape as input for different packages.

```python
data=np.array([[1,2],[3,4]])
print(data)
print(data.shape,'\n')

data = np.reshape(data,(4,1))
print(data)
print(data.shape,'\n')

data = np.reshape(data,(1,4))
print(data)
print(data.shape,'\n')
```

```python
# One can reshape using either numpy built-in function like above,
# or the .reshape() Method assocaited a numpy array object.

data=np.array([[1,2],[3,4]])
print(data)
print(data.shape,'\n')

data = data.reshape(4,1)
print(data)
print(data.shape,'\n')
```

```python
# Note about the differences between the following arrays:
data = np.array([1,2,3,4])
print(data)
print(data.shape,'\n')

data = np.array([[1,2,3,4]])
print(data)
print(data.shape,'\n')

data = data.reshape(4,1)
print(data)
print(data.shape,'\n')
```

```python
data = np.array([[1,2],[3,4]])
print(data)
print(data.shape,'\n')

data = data.flatten()
print(data)
print(data.shape,'\n')

# Note: One can use the value "-1" as an input argument,
# for an automatic determination of a matrix/vector dimension.

#data = np.reshape(data,(-1,1))
data = data.reshape(-1,1)

print(data)
print(data.shape,'\n')
```

```python
# Stacking arrays:
data=np.arange(5)
print(data)
print(data.shape,'\n')

data=np.reshape(data, (1,-1))
print(data.shape,'\n')

datah = np.hstack((data,data,data))
print(datah)
print(datah.shape,'\n')

datav = np.vstack((data,data,data))
print(datav)
print(datav.shape)
```

## Indexing and Slicing
The rule is the same as that in Python list.

```python
a=np.arange(0,6,1)
print(a)
print(a[3:5])
print(a[-4:-1])

# Slicing syntax:
# array_name[start : end : step]

print(a[0:-1:2])
print(a[-1:0:-2])
```

## Vectorized Expressions

```python
x=np.array([1,2,3,4])
y=np.array([5,6,7,8])
print(x+y)
print(x*y)
print(x**2)
```

```python
# For plotting, remember to import the following package first:
import matplotlib.pyplot as plt

x=np.linspace(0,2*np.pi,101)

# np.sin and np.cos are Numpy built-in functions, which act on Numpy arrays.
# Be careful about whether a function acts on an arrary directly, or just on a single number.

y1=np.sin(x)
y2=np.cos(x)

plt.plot(x,y1,label='sin(x)')
plt.plot(x,y2,label='cos(x)')

plt.xlabel('x',fontsize=16)
plt.ylabel('y',fontsize=16)

plt.legend()

plt.show()
```

```python
# Define our own function
def heaviside(x):
    if x > 0:
        return 1
    else:
        return 0
    
print(heaviside(-1))
print(heaviside(3.0))
```

```python
x=np.linspace(-5,5,101)

# Note:: The following command alone will create errors: heaviside(x)
# Our function alone is not well-defined for a numpy array; 
# it also returns only a single value.

heaviside = np.vectorize(heaviside)
# Now the function can take a numpy array input.
y = heaviside(x)

plt.title('Heaviside Function')
plt.xlabel('x',fontsize=16)
plt.ylabel('y',fontsize=16)

plt.plot(x,y)
plt.show()
```

```python
# Define heaviside function directly for an numpy array:
def hside(xarray):
    harray = (xarray > 0).astype(np.int32)
    return harray

x=np.linspace(-5,5,101)
y=hside(x)

plt.title('Heaviside Function')
plt.xlabel('x',fontsize=16)
plt.ylabel('y',fontsize=16)

plt.plot(x,y)
plt.show()
```

```python
# Aside: Another way to construct a Heaviside function with Boolean operation.

x = np.array([-1,0,1,2,3])
print(x > 0,'\n')

print((x > 0).astype(np.int32))
```

## Numpy Linear Algebra Packages
The NumPy linear algebra functions in **numpy.linalg** provide efficient low level implementations of standard linear algebra algorithms. You may check online the documentation page for further information:
https://numpy.org/doc/stable/reference/routines.linalg.html

```python
A = np.random.rand(2,2)
B = np.random.rand(2,2)

# Matrix multiplication
print(np.matmul(A,B),'\n')

# Raise a square matrix to the (integer) power n
n = 3
print(np.linalg.matrix_power(A, n),'\n')

# Compute the eigenvalues and right eigenvectors of a square array
C = np.array([[1,2],[2,1]])

# Assign the returned eigenvalues and eigenvectors to two variables:
val, vec = np.linalg.eig(C)
print(val)
print(vec)

# For further information, check:
# help(np.linalg.eig)
```
