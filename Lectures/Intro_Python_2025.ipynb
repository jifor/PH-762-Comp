## Dictionaries

Dictionary consists of **keys** (numbers or strings) and corresponding **values** (any object types). The popular file format JSON is basically a text file containing python dictionaries. 

```python
dic_atomic_num = {'H': 1, 'He': 2, 'Li': 3}

# We extract the keys and values using the "list" Built-in Function,
# which convert the results to lists:
keys = list(dic_atomic_num.keys())
values = list(dic_atomic_num.values())

print(keys)
print(values)

# The get Method:
print(dic_atomic_num.get('H'))
print(dic_atomic_num.get('He'))
print(dic_atomic_num.get('Li'))
```

```python
# Add new elements to the dictionary:
dic_atomic_num['Be']=4
dic_atomic_num['B']=5
dic_atomic_num['C']=6

# Remove an existing element:
dic_atomic_num.pop('C')

keys = list(dic_atomic_num.keys())
values = list(dic_atomic_num.values())

print(keys)
print(values)
```

### Break and Continue

The **break** statement breaks out of a "for" or "while" loop, and moves onto the next section of code.

The **continue** statement skips the current iteration of a "for" or "while" loop, and moves onto the next iteration.

```python
for ii in range(10):
    
    if ii == 5: break
        
    print(ii)
```

```python
for ii in range(10):
    
    if ii == 5: continue
        
    print(ii)
```

**Exercise 2.** For a given string (e.g. 'Hello World'), write a loop to print the characters at an even index number (including 0). The **modulo operator %** will be useful here.

```python
# Your code starts here:
string = 'Hello World'

for i in range(len(string)):
    if i % 2 == 0:
        print(string[i])
```

**Exercise 3.** Re-rewrite the above code, by making it a function with the string as an input argument, and returning the desired characters as a Python list.

```python
# Your code starts here:
def parse(string):
    letters = []

    for i in range(len(string)):
        if i % 2 == 0:
            letters.append(string[i])

    return letters

parse('Hello World')
```
