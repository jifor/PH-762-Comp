# Trigonometric Interpolation

## Fourier Series
For a periodic function $f(x)$ with $x\in [0, 2\pi]$, **we do not know about the actual functional form of $f(x)$**. But, we do know about the values of $(x_j, y_j \equiv f(x_j))$, for $j=0, 1, 2, ..., n-1$.

Our goal is to produce an interpolated function $p(x)$ for $x\in [0, 2\pi]$, using $(x_j, y_j)$. In terms of the Fourier Series, the answer is:

$p(x) = \frac1{2}a_0 +\sum_{k=1}^{m-1} [a_k \cos(kx) + b_k \sin(kx)] +\frac1{2}a_m \cos(mx)$.

For an arbitrary distribution of $x_j$, $a_k$ and $b_k$ can be obtained by solving a linear equation.

In the special case of an equally spaced grid, $x_j = 2\pi \frac{j}{n}, j =0,1,...,n-1$, then $a_k$ and $b_k$ are directly computed as:

$a_k = \frac1{m} \sum_{j=0}^{n-1} y_j \cos(kx_j)$, for $k=0,1, ...,m$.

$b_k = \frac1{m} \sum_{j=0}^{n-1} y_j \sin(kx_j)$, for $k=1,2, ..., m-1$.

**Note:** Here, we consider $n$ to be an even number, and $m\equiv n/2$.

```python
# Import numpy and matplotlib for scientific calculation/visualization.
import numpy as np
import matplotlib.pyplot as plt
```

```python
from math import pi

def f(x):
    #return np.sin(2*x) + 0.5*np.cos(x)
    return np.exp(np.sin(2*x))

def generatedata(n,f):
    datax = 2*pi*np.arange(n)/n
    datay = f(datax)
    return datax, datay

def computeparams(datax,datay):
    n=datax.size
    m=int(n/2)
    
    aparams = np.zeros(m+1)
    
    #Note: we still define the same array size for bparams:
    bparams = np.zeros(m+1)

    for k in range(0, m+1):
        aparams[k] = np.dot(datay, np.cos(k*datax))/m

    #Note about the array range:
    for k in range(1, m):
        bparams[k] = np.dot(datay, np.sin(k*datax))/m

    return aparams, bparams

def triginterp(aparams, bparams, x):
    n = aparams.size + bparams.size
    m = int(n/2)
    
    #The first and last terms in the series expansion:
    val = 0.5*(aparams[0] + aparams[-1]*np.cos(m*x))
    
    for k in range(1,m):
        val = val + aparams[k]*np.cos(k*x) + bparams[k]*np.sin(k*x)

    return val
```

```python
nplot = 101
x = np.linspace(0,2*pi,nplot)
y = f(x)

fig, ax = plt.subplots(figsize=(6,4))
ax.plot(x,y, '-k', label='f(x)')

# Here, we want n to be even.
n = 10
datax, datay = generatedata(n, f)

ax.plot(datax, datay, 'sr', label='raw data', markersize=10)

aparams, bparams = computeparams(datax, datay)
pofx = triginterp(aparams, bparams, x)
ax.plot(x,pofx,'--bo',label='p(x)', alpha=0.5)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
```

```python
# Print out the a and b coefficients:
print('as:', aparams,'\n')
print('bs:', bparams)
```

## Discrete Fourier Transform (DFT)
We consider an equally spaced grid: $x_j = 2\pi \frac{j}{n}, j =0,1,...,n-1$, with $n \equiv 2m$ being an even number.

Given $(x_j, y_j)$, we want an interpolated function $p(x)$ with $p(x_j)=y_j$. In terms of complex exponentials, the anser is:

$p(x) = \sum_{k=-m+1}^{k=m-1} c_k e^{ikx}+c_{-m}\cos(mx)$.

As the complex exponentials are orthogonal to each other also in the discrete case, the coefficients $c_ks$ can be computed directly as:

$c_k = \frac1{n}\sum_{j=0}^{j=n-1} y_j e^{-ikx_j}$, for $k= -m, -m+1, ..., m-2, m-1$.

We can define a shifted version of DFT:

$\tilde y_k = \sum_{j=0}^{j=n-1} y_j e^{-ikx_j}$, for $k= 0, 1, 2, ..., n-1$.

The corresponding inverse DFT is:

$y_j = \frac1{n}\sum_{k=0}^{k=n-1} \tilde y_k e^{ikx_j}$, for $j = 0, 1, 2, ..., n-1$.


## Fast Fourier Transform (FFT)
We consider an equally spaced grid: $x_j = 2\pi \frac{j}{n}, j =0,1,...,n-1$, with $n \equiv 2m \equiv 2^p$ being an even number. Here, $p$ is a non-negative integer.

Given $(x_j, y_j)$, we want an interpolated function $p(x)$ with $p(x_j)=y_j$. In terms of the shifted version of DFT, the answer is:

$p(x) = \frac1{n}\sum_{k=0}^{k=m-1} \tilde y_k e^{ikx}+ \frac1{n} \tilde y_m \cos(mx) + \frac1{n} \sum_{k=m+1}^{k=n-1} \tilde y_k e^{i(k-n)x}$.

In FFT, $\tilde y_k$ is computed as

$y_k = \sum_{l=0}^{l=m-1} y_{2l}e^{-i 2\pi kl/m} + e^{-i2\pi k/n} \sum_{l=0}^{l=m-1} y_{2l+1}e^{-i2\pi kl/m} \equiv \tilde y^{even}_{k} + e^{-i2\pi k/n} \tilde y^{odd}_{k}$.

The above **divide-and-conquer** procedue can be carried out recursively until $n=1$ ($\tilde y_0 = y_0$).






```python
# Sample FFT Code:
from math import pi

def f(x):
    #return np.sin(2*x) + 0.5*np.cos(x)
    return np.exp(np.sin(2*x))

def generatedata(n,f):
    datax = 2*pi*np.arange(n)/n
    datay = f(datax)
    return datax, datay

def fft(ys):
    n = ys.size
    m = n//2 #divide with integral result (discard remainder)

    if n == 1:
        ytils = ys
    else:
        evens = fft(ys[::2])
        odds = fft(ys[1::2])
        coeffs = np.exp(-1j*2*pi*np.arange(m)/n)
        first = evens + coeffs*odds
        second = evens - coeffs*odds
        ytils = np.concatenate((first, second))
    
    return ytils

#def fftinterp(ytils, x):
def fftinterp(x):
    n = ytils.size
    m = n//2 #divide with integral result (discard remainder)
    val = ytils[:m]@np.exp(1j*np.arange(m)*x)
    val += ytils[m]*np.cos(m*x)
    val += ytils[m+1:]@np.exp(1j*np.arange(-m+1,0)*x)
    #Note: "@" operator is for dot product of 1D arrays, without the need of explicit loops and indices.

    return val/n

# Here, we want n to be even and = 2^p.
n = 8
datax, datay = generatedata(n, f)
ytils = fft(datay)

# The fftinterp function we defined takes one single value of x at a time:
x = 0.3
pofx = fftinterp(x)
print("x, p(x)=", x,pofx.real)

# Vectorize the fuction, so that it can take directly an numpy array as input:
fftinterp = np.vectorize(fftinterp)
nplot = 101
x = np.linspace(0,2*pi,nplot)
pofx = fftinterp(x)

fig, ax = plt.subplots(figsize=(6,4))
y = f(x)
ax.plot(x,y, '-k', label='f(x)')
ax.plot(datax, datay, 'sr', label='raw data', markersize=10)
ax.plot(x,pofx.real,'--bo',label='Re[p(x)]', alpha=0.5)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
```
