```python
import numpy as np
import matplotlib.pyplot as plt

# define constants
hbar = 1.054571817e-34 # J⋅s
e = 1.60218e-19 # C
eps_0 = 8.8541878188e-12 # C2⋅kg−1⋅m−3⋅s2
m_e = 9.1093837e-31 # kg
a = 0.529e-10 # m
```

```python
# define the energy function
def E(n,Joules=True):
    energy = -(m_e / (2 * hbar**2) * (e**2 / (4 * np.pi * eps_0))**2) / n**2
    if Joules == True:
        return energy # J
    else:
        return energy / 1.60218e-19 # conversion from J to eV

def f(w, r, n, l):
    u = w[0]
    v = w[1]
    energy = E(n)

    alpha = (2 * m_e * e**2) / (4 * np.pi * hbar**2 * eps_0)
    beta = 2 * m_e * energy / hbar**2
    K = l*(l+1) / r**2 - alpha / r - beta

    fu = v
    fv = K * u

    return np.array([fu,fv])
```

```python
# quantum numbers
n = 1
l = 0

# discretize r
r_i = a/10
r_f = 25 * a
n_steps = 1001
h = (r_f - r_i) / n_steps

r = np.linspace(r_i,r_f,n_steps)

u = np.zeros(n_steps)
u[0] = 0
v = np.zeros(n_steps)
v[0] = 1
w = np.array([u[0],v[0]])

# implement RK-4
for i in range(n_steps-1):
    k1 = f(w,r[i],n,l)
    k2 = f(w+k1*h/2, r[i]+h/2,n,l)
    k3 = f(w+k2*h/2, r[i]+h/2,n,l)
    k4 = f(w+k3*h, r[i]+h,n,l)

    w = w + h*(k1+2*k2+2*k3+k4)/6

    u[i+1] = w[0]
    v[i+1] = w[1]

plt.plot(r/a,u/r,label='RK4')

def Psi_100(r):
    return 2 / a**(1.5) * np.exp(-r/a)

plt.plot(r/a, Psi_100(r),label='actual soln')
plt.legend()
```
