```python
from scipy import constants as const
import numpy as np
import matplotlib.pyplot as plt
```

```python
# constants
e = const.e
hbar = const.hbar
eps_0 = const.epsilon_0
m_e = const.m_e
pi = const.pi
a = 4 * pi * eps_0 * hbar**2 / m_e / e**2
print(a)

# define the potential
def V(r, l):
    alpha = e**2 / (4 * pi * eps_0)
    beta = (hbar**2 / 2 / m_e) * l * (l+1)
    return -alpha / r + beta / r**2

# Define function for Simpson's rule
def Simpson(psi, a, b, n):
    if n % 2 == 0:
        print('Error: n must be an odd number!')
    else:
        h = (b-a)/(n-1)
        x = np.linspace(a,b,n)

        # Create array of c_i
        c = np.ones(n)
        for i in range(len(c)):
            if i % 2 == 0:
                c[i] = 2
            else:
                c[i] = 4
        c[0], c[-1] = 1, 1

        contribution = (h/3)*c*psi
        return np.sum(contribution)
```

```python
# create a function for plotting the actul results
def Actual_Radial_Func(r,n,l):
    expo = np.exp(-r/(n*a))
    if n == 1:
        return 2 * a**(-1.5) * expo
    if n == 2:
        if l == 0:
            poly = 1 - r / (2*a)
            return (1/np.sqrt(2)) * a**(-1.5) * poly * expo
    if n==3:
        if l==2:
            const = 4 * (a**-1.5) / (81 * np.sqrt(30))
            poly = (r/a)**2
            return const * poly * expo
```

```python
# quantum numbers
l = 0

# discretize the r-grid
r_i = 0
r_f = 100*a
n = 1001
dx = (r_f-r_i) / (n-1)
k = hbar**2 / (2 * m_e * dx**2)
print(k)

r = np.linspace(r_i,r_f,n)
r_plot = r[1:-1]

def Solve(l=0):
    '''
    function that will solve the Hamiltonian eigenvalue problem, 
    normalize the eigenvectors, and return the energies and normalized
    vectors
    '''
    # build the hamiltonian matrix
    H = np.zeros((n-2,n-2))
    
    H[0][0] = 2*k + V(r[1],l)
    H[0][1] = -k
    
    H[-1][-2]= -k
    H[-1][-1]= 2*k + V(r[n-2],l)
    
    for ii in range(1,n-3):
        H[ii][ii] = 2*k + V(r[ii+1],l)
        H[ii][ii+1] = -k
        H[ii][ii-1] = -k

    # solve the Hamiltonian eigenenergies and vectors
    vals, vecs = np.linalg.eigh(H)
    energies = vals[0:3]/e # return the first three energies in eV

    # add in the boundary conditions to the first three vectors
    wavefuncs = np.zeros((n-2,3))
    wavefuncs = vecs[:,0:3]

    # # normalize the wavefunctions in wavefuncs
    # for i in range(3):
    #     psi = wavefuncs[:,i]
    #     I = Simpson(psi**2,r_i,r_f,n)
    #     c = 1 / np.sqrt(I)
    #     wavefuncs[:,i] = c * psi 

    # return the first three energies and vectors
    return energies, wavefuncs

# define a function that will normalize the outputted wave function
def Normalization(psi):
    # create the radial function from u
    radial = psi / r_plot
    I = Simpson(radial**2,r_i,r_f,n-2)
    c = 1 / np.sqrt(I)
    return c * radial
```

```python
# Solve for the eigenenergies and vectors
energies, psis = Solve()
norm_radial = np.zeros(psis.shape)
for i in range(3):
    norm_radial[:,i] = Normalization(psis[:,i])
    plt.plot(r_plot/a,-norm_radial[:,i],label=f'{energies[i]}')
plt.legend()
plt.axhline(y=0, ls='--', c='r')
plt.xlim(0,25)
```

# Angular part

```python
def Y_22(phi):
    return np.sqrt(15/(32*pi)) * np.cos(2*phi)

rlist = np.linspace(0,20*a,n)
philist = np.radians(np.arange(0,361,1))
rmesh,phimesh = np.meshgrid(rlist,philist)

Psi_322 = Actual_Radial_Func(rmesh,3,2) * Y_22(phimesh)
print(Psi_322.shape)

fig,ax = plt.subplots(dpi=120,subplot_kw=dict(projection='polar'))
ax.contourf(phimesh,rmesh,Psi_322,100,cmap='plasma')
```

```python
final_n = 250
rlist = np.linspace(0,25*a,final_n)
philist = np.radians(np.arange(0,361,1))
rmesh,phimesh = np.meshgrid(rlist,philist)
R_30 = psis_2[:final_n,0]

R_matrix = np.zeros((361,final_n))
for i in range(361):
    R_matrix[i,:] = R_30

fig,ax = plt.subplots(dpi=120,subplot_kw=dict(projection='polar'))
ax.contourf(phimesh,rmesh,R_matrix,100)
# ok that's what I expect to see now baby
```

```python
# discretize the phi grid
phi_0 = 0
phi_f = 2*pi
n = 1001
h = (phi_f - phi_0) / (n-1)
phi = np.linspace(phi_0,phi_f,n)

def f(r,phi,m):
    Phi = r[0]
    P = r[1]
    fphi = P
    fP = -m**2 * phi
    return np.array([fphi,fP])

Phi = np.zeros(n)
P = np.zeros(n)
Phi[0] = 0
P[0] = 0

r = np.array([Phi[0],P[0]])

m = 2
for i in range(n-1):
    k1 = f(r,phi[i],m)
    k2 = f(r+k1*h/2,phi[i]+h/2,m)
    r = r + k2 * h
    Phi[i+1] = r[0]
    P[i+1] = r[1]

plt.plot(phi,Phi)
```
